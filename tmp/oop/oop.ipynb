```python
class Temperatuur:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    # @celsius.setter
    def celsius(self, waarde):
        if waarde < -273.15:
            raise ValueError("Te koud")
        self._celsius = waarde

    def __eq__(self, other):
        return True



t = Temperatuur(20)
t.celsius(22)
# t.celsius

t=="bla"
```
